```python
data= pd.read_csv('/kaggle/input/datasets/souradippal/fridge-price-prediction-dataset/fridge_price_predictor_dataset_real_brands.csv')
print ((data.duplicated().sum()))

print ('########')
data.head()
```

splet obj & num

```python
obj_data=data.select_dtypes(include = object)
num_data=data.select_dtypes(exclude = object)
```

filling null data

```python
clean_obj =obj_data.fillna(obj_data.mode())
clean_obj =clean_obj.fillna('no_value')
clean_num =num_data.fillna(num_data.mean())
```

```python
clean_obj.info()
clean_num.info()
```

# concat our data

```python
new_data= pd.concat([clean_obj,clean_num],axis=1)
new_data.head()
```

```python
new_data.info()
```

# data processing obg to num

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
obj_names= clean_obj.columns
for i in obj_names:
    new_data[i]= le.fit_transform(new_data[i])
```

```python
new_data.head()
```

```python
new_data.corr()
```

# Split


```python
x= new_data.drop(['Price'], axis=1)
y= new_data['Price']
```

# training

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x,y)
```

```python
lr.score(x,y)
```
